# V4 CPU data and split audit

This is a thin Kaggle wrapper. Attach an immutable V4 package/config and run the local audit CLI. It does not train, tune thresholds, or submit another job.

```python
import os
import subprocess
from pathlib import Path

config = Path(os.environ['RV4_CONFIG'])
subprocess.run([
    'python', '-m', 'rescuevision.v4.cli', 'audit',
    '--config', str(config),
], check=True)
```
