# V4 final inference bundle verification

This notebook verifies the bundle manifest and checkpoint hash before inference. It does not tune thresholds, replace a corrupt artifact, or publish a model.

```python
import json
import os
from pathlib import Path

from rescuevision.v4.bundle import load_inference_bundle

bundle_dir = Path(os.environ['RV4_BUNDLE_DIR']).resolve()
bundle = load_inference_bundle(bundle_dir)
print(json.dumps({
    'stage': 'L9_bundle_verify',
    'bundle': str(bundle_dir),
    'model_id': bundle.spec.model_id,
    'model_revision': bundle.spec.model_revision,
    'checkpoint_sha256': bundle.checkpoint_sha256,
    'preprocessing_identity': bundle.metadata['preprocessing_identity'],
}, indent=2))
```
