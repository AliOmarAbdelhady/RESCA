# V4 DINOv3 teacher profile preflight

This notebook only records the runtime profile for the approved DINOv3 teacher configuration. It does not download the gated 7B checkpoint, extract held-out features, or train an optimizer. Set `RV4_CONFIG` to an immutable package configuration before running.

```python
import json
import os
import subprocess
import sys
from pathlib import Path

config = Path(os.environ['RV4_CONFIG']).resolve()
result = subprocess.run([
    sys.executable, '-m', 'rescuevision.v4.cli', 'profile',
    '--config', str(config),
], check=True, capture_output=True, text=True)
profile = json.loads(result.stdout)
print(json.dumps({'stage': 'L7_profile_preflight', 'config': str(config), 'profile': profile}, indent=2))
```
