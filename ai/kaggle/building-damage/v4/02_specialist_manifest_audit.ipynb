# V4 specialist manifest audit

This notebook validates post-only and visible-defect JSONL manifests through the immutable V4 adapters. It preserves grouped splits and missing annotations; it does not create pseudo-masks or train a specialist.

```python
import json
import os
from pathlib import Path

from rescuevision_ai.v4.specialist_datasets import load_defect_manifest, load_post_only_manifest

post_only_path = Path(os.environ['RV4_POST_ONLY_MANIFEST']).resolve()
defect_path = Path(os.environ['RV4_DEFECT_MANIFEST']).resolve()
post_only = load_post_only_manifest(post_only_path)
defects = load_defect_manifest(defect_path)
print(json.dumps({
    'stage': 'L8_manifest_audit',
    'post_only_manifest': str(post_only_path),
    'post_only_records': len(post_only),
    'defect_manifest': str(defect_path),
    'defect_records': len(defects),
}, indent=2))
```
